```python
import re
import unicodedata

import pandas as pd
```

```python

def clean_column(name):
    # Normalize weird characters to ASCII
    name = unicodedata.normalize('NFKD', name).encode('ascii', 'ignore').decode('utf-8')
    # Lowercase, strip, replace spaces/specials with underscores
    name = re.sub(r'[^a-zA-Z0-9]+', '_', name.strip().lower())
    # Remove trailing underscores
    return re.sub(r'_+$', '', name)
```

```python
raw_df = pd.read_excel(
    '/home/tomf/Coding/Freehily/Stronger/stronger/backend/databases/exercises/Functional+Fitness+Exercise+Database+(version+2.9).xlsx',
    sheet_name='Exercises', skiprows=15).iloc[:, 1:] 
```

```python
raw_df.to_csv('exercise_data_raw.csv', index=False)
```

```python
# Apply to columns
raw_df.columns = [clean_column(col) for col in raw_df.columns]
```
